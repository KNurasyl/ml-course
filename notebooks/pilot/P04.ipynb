## Разобранный пример

Строка матрицы содержит признаки одного примера. Вектор весов содержит по одному коэффициенту на признак. Матрица (n,d), умноженная на вектор (d,), даёт n прогнозов. Число примеров и число признаков выполняют разные роли.

### Вычислить скалярное произведение

```python
a = np.array([1., 2.]); b = np.array([3., 4.])
print("Вручную:", 1*3 + 2*4, "NumPy:", a @ b)
assert a @ b == 11
```

### Проверить матрицу и веса

```python
X = np.array([[1., 2., 3.], [4., 5., 6.]])
w = np.array([1., 0., -1.])
print("X:", X.shape, "w:", w.shape, "ответ:", (X @ w).shape)
np.testing.assert_array_equal(matvec(X, w), [-2, -2])
display(pd.DataFrame(X, columns=["признак 1", "признак 2", "признак 3"]))
try:
    matvec(X, [1, 2])
except ValueError as error:
    print("Ожидаемая ошибка:", error)
```
