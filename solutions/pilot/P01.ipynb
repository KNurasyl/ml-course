## Разобранный пример

Рассмотрим искусственные пары (1,3), (2,5), (3,7). Формула y = wx + b задаёт семейство моделей. При w=2 и b=1 она совпадает с этими тремя ответами. Это удобный учебный случай, а не доказательство надёжности на реальных данных.

### Вычислить прогнозы

```python
x = np.array([1., 2., 3.]); y = np.array([3., 5., 7.])
w, b = 2., 1.
prediction = w * x + b
display(pd.DataFrame({"x": x, "известный y": y, "предсказание": prediction}))
assert np.array_equal(prediction, y)
print("Новый вход x=4:", w * 4 + b)
```

### Связать формулу с графиком

```python
fig, ax = plt.subplots()
grid = np.linspace(0, 4, 50)
ax.scatter(x, y, color="#1f5a99", label="Учебные примеры")
ax.plot(grid, w * grid + b, color="#333333", label="y = 2x + 1")
ax.set(xlabel="Вход x, условные единицы", ylabel="Ответ y, условные единицы", title="Простая линейная модель")
ax.legend(); fig.tight_layout()
fig.savefig(FIGURES / "P01_model.png", dpi=130)
plt.show()
```
